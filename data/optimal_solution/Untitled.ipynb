```python
import pandas as pd
from gnncloudmanufacturing.validation import total_cost_from_gamma
from gnncloudmanufacturing.data import read_fatahi_dataset
from gnncloudmanufacturing.utils import delta_from_gamma
import json
import numpy as np
import os
```

```python
dataset = read_fatahi_dataset('../fatahi.xlsx')
optimal = pd.read_csv('dp.csv', index_col=0)
```

```python
solution = optimal.solution.to_dict()
problem = dataset[0]
```

```python
for problem in dataset:
    problem_solution = eval(solution[problem['name']].replace("'", "\""))
    n_tasks, n_operations, n_cities = problem['name'].split(',')
    n_tasks, n_operations, n_cities = int(n_tasks), int(n_operations), int(n_cities.split('-')[0])

    gamma = np.zeros((n_operations, n_tasks, n_cities))
    for t in problem_solution:
        for o in problem_solution[t]:
            c = problem_solution[t][o]
            i = int(t[9:]) - 1
            j = int(o[13:]) - 1
            k = int(c[4:]) - 1
            gamma[j, i, k] = 1
    delta = delta_from_gamma(problem, gamma)
    
    if not os.path.exists(problem["name"]):
        os.makedirs(problem["name"])
    np.save(f'{problem["name"]}/gamma.npy', gamma)
    np.save(f'{problem["name"]}/delta.npy', delta)
    print(f'{problem["name"]}: {total_cost_from_gamma(problem, gamma, delta):.4f}')
```
