```python
from gnncloudmanufacturing.data import read_fatahi_dataset, sample_dataset
from gnncloudmanufacturing.dp_solver import dp_solve
from gnncloudmanufacturing.utils import delta_from_gamma
from tqdm.auto import tqdm
import numpy as np
import os
```

```python
dataset = read_fatahi_dataset('../../../data/fatahi.xlsx')
```

```python
for problem in [dataset[0], dataset[3], dataset[6], dataset[9], dataset[12], dataset[15]]:
    n_tasks, n_operations, n_cities = problem['n_tasks'], problem['n_operations'], problem['n_cities']
    dirname = f'{n_tasks}-{n_operations}-{n_cities}'
    if not os.path.exists(dirname):
        os.makedirs(dirname)
    np.save(f'{dirname}/dist.npy', problem['dist'])
    np.save(f'{dirname}/op_cost.npy', problem['op_cost'])
    np.save(f'{dirname}/productivity.npy', problem['productivity'])
    np.save(f'{dirname}/time_cost.npy', problem['time_cost'])
```

```python
for problem in [dataset[0], dataset[3], dataset[6], dataset[9], dataset[12], dataset[15]]:
    n_tasks, n_operations, n_cities = problem['n_tasks'], problem['n_operations'], problem['n_cities']
    dirpath = f'{n_tasks}-{n_operations}-{n_cities}/'
    train_dataset = sample_dataset(
        100, 
        [n_tasks, n_tasks], 
        [n_operations, n_operations], 
        [n_cities, n_cities], 
        dirpath=dirpath,
        random_seed=0
    )
    for problem in tqdm(train_dataset):
        gamma = dp_solve(problem)
        delta = delta_from_gamma(problem, gamma)
        if not os.path.exists(f'{dirpath}/{problem["name"]}'):
            os.makedirs(f'{dirpath}/{problem["name"]}')
        np.save(f'{dirpath}/{problem["name"]}/delta.npy', delta)
        np.save(f'{dirpath}/{problem["name"]}/gamma.npy', gamma)
```
