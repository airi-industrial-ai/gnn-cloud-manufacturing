```python
from gnncloudmanufacturing.data import read_fatahi_dataset
from gnncloudmanufacturing.dp_solver import dp_solve
from gnncloudmanufacturing.validation import total_cost_from_gamma
from gnncloudmanufacturing.utils import delta_from_gamma

from time import time
import pandas as pd
from tqdm.auto import tqdm
```

```python
dataset = read_fatahi_dataset('../data/fatahi.xlsx')
for problem in dataset:
    print(f'Problem: {problem["name"]}')
```

```python
problem_name = []
total_cost = []
comp_time = []
for problem in tqdm(dataset):
    start = time()
    gamma = dp_solve(problem)
    delta = delta_from_gamma(problem, gamma)
    total = total_cost_from_gamma(problem, gamma, delta)
    total_cost.append(total)
    problem_name.append(problem['name'])
    comp_time.append(time() - start)
```

```python
pd.DataFrame({'problem_name': problem_name, 'total_cost': total_cost, 'comp_time': comp_time}).round(2)
```

```python
pd.DataFrame({'problem_name': problem_name, 'total_cost': total_cost, 'comp_time': comp_time}).round(2).to_excel('dp_results.xlsx')
```
